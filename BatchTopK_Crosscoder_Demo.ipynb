## Prepare Inputs for KL-Divergence (First 5 Prompts)

### Subtask:
Tokenize the first 5 refusal prompts and extract layer 13 activations from `loaded_base_model` and `loaded_chat_model`.


**Reasoning**:
To extract activations for the first 5 refusal prompts, I will iterate through the prompts, tokenize them, and use `nnsight`'s tracing capabilities to get the activations from the specified layer for both models, storing them in separate lists.



````python
import torch as th
from tqdm.auto import tqdm

base_activations = []
chat_activations = []
layer = 13

print(f"Extracting activations for the first 5 prompts from layer {layer}...")

for i, prompt_data in tqdm(enumerate(refusal_prompts[:5]), total=5, desc="Processing prompts"):
    instruction = prompt_data['instruction']
    conversation = [{'role': 'user', 'content': instruction}]

    tokens = loaded_chat_model.tokenizer.apply_chat_template(conversation, return_tensors="pt").to(cc_device)

    with loaded_base_model.trace(tokens):
        base_acts = get_layer_output(loaded_base_model, layer).save()
    base_activations.append(base_acts.cpu())

    with loaded_chat_model.trace(tokens):
        chat_acts = get_layer_output(loaded_chat_model, layer).save()
    chat_activations.append(chat_acts.cpu())

print(f"\nCollected {len(base_activations)} base model activations.")
print(f"Collected {len(chat_activations)} chat model activations.")

if base_activations:
    print(f"Shape of first base activation: {base_activations[0].shape}")
    print(f"Shape of first chat activation: {chat_activations[0].shape}")
````

## Generate Crosscoder Latent Activations

### Subtask:
Generate baseline latent representations for each prompt using the `crosscoder_model` from the previously extracted `base_activations` and `chat_activations`.


**Reasoning**:
To generate baseline latent representations, I will iterate through the collected base and chat activations, prepare the input tensor for the `crosscoder_model` by reshaping, moving to the correct device, and stacking, then pass it through the model and store the output.



````python
crosscoder_outputs = []

print("Generating crosscoder latent activations...")

for base_act, chat_act in tqdm(zip(base_activations, chat_activations), total=len(base_activations), desc="Processing activations for crosscoder"):
    # Reshape from [1, seq_len, d] to [seq_len, d]
    base_act_reshaped = base_act.squeeze(0).to(cc_device)
    chat_act_reshaped = chat_act.squeeze(0).to(cc_device)

    # Stack along a new dimension to get [seq_len, 2, d]
    cc_input = th.stack([base_act_reshaped, chat_act_reshaped], dim=1).float()

    # Pass through the crosscoder model
    cc_output = crosscoder_model(cc_input)

    # Append to the list
    crosscoder_outputs.append(cc_output.cpu())

print(f"\nCollected {len(crosscoder_outputs)} crosscoder latent outputs.")
if crosscoder_outputs:
    print(f"Shape of first crosscoder output: {crosscoder_outputs[0].shape}")
````

## Compute baseline log probabilities for KL-Divergence

### Subtask:
For each of the first 5 refusal prompts, compute the baseline log probabilities from both the `loaded_base_model` and `loaded_chat_model` for KL-Divergence calculations.


**Reasoning**:
To compute KL-Divergence, I need the log probabilities from the language modeling head of both the base and chat models for each refusal prompt. I will use `nnsight`'s tracing capabilities to extract the `lm_head` output (logits), apply `log_softmax` to get log probabilities, and store them in separate lists for the first 5 prompts.



````python
import torch.nn.functional as F

base_log_probs = []
chat_log_probs = []

print("Computing baseline log probabilities for the first 5 prompts...")

for i, prompt_data in tqdm(enumerate(refusal_prompts[:5]), total=5, desc="Processing prompts for log probs"):
    instruction = prompt_data['instruction']
    conversation = [{'role': 'user', 'content': instruction}]

    tokens = loaded_chat_model.tokenizer.apply_chat_template(conversation, return_tensors="pt").to(cc_device)

    with loaded_base_model.trace(tokens):
        base_logits = loaded_base_model.lm_head.output.save()
    base_log_probs.append(F.log_softmax(base_logits.cpu(), dim=-1))

    with loaded_chat_model.trace(tokens):
        chat_logits = loaded_chat_model.lm_head.output.save()
    chat_log_probs.append(F.log_softmax(chat_logits.cpu(), dim=-1))

print(f"\nCollected {len(base_log_probs)} base model log probability tensors.")
print(f"Collected {len(chat_log_probs)} chat model log probability tensors.")

if base_log_probs:
    print(f"Shape of first base log probability tensor: {base_log_probs[0].shape}")
    print(f"Shape of first chat log probability tensor: {chat_log_probs[0].shape}")
````

## Compute KL-Divergence Attribution - Setup

### Subtask:
Prepare the necessary components for computing KL-Divergence attribution, including defining a KL-Divergence function and initializing storage for attribution scores.


**Reasoning**:
To prepare for KL-Divergence attribution, I will define a helper function for KL-Divergence calculation, initialize a list to store the attribution scores, and determine the total number of latents from the `crosscoder_outputs` to facilitate iteration.



````python
def compute_kl_divergence(log_P, log_Q):
    """
    Computes the KL-Divergence between two log probability distributions.
    Args:
        log_P (torch.Tensor): Log probabilities of the reference distribution (e.g., chat model output). Shape: [batch, seq_len, vocab_size]
        log_Q (torch.Tensor): Log probabilities of the intervention distribution. Shape: [batch, seq_len, vocab_size]
    Returns:
        torch.Tensor: Scalar KL-Divergence value, averaged over tokens and batch.
    """
    # Ensure inputs are log probabilities (already log_softmaxed)
    # kl_div expects log_P and P or log_Q and Q. Here we have log_P and log_Q.
    # F.kl_div(input, target, ...)
    # input is log-probability (log_Q here), target is probability (exp(log_P) here)
    kl_div_per_token = F.kl_div(log_Q, F.softmax(log_P, dim=-1), reduction='none')
    # Sum over vocabulary, then average over sequence length and batch
    return kl_div_per_token.sum(dim=-1).mean()

# Initialize storage for KL-Divergence attribution scores
kl_divergence_attributions = []

# Determine the total number of latents from crosscoder_outputs
# crosscoder_outputs is a list of tensors, each with shape [seq_len, 2, n_latents]
# We need the last dimension of the last tensor's shape (assuming all have same shape)
if crosscoder_outputs:
    num_latents = crosscoder_outputs[0].shape[-1]
    print(f"Defined compute_kl_divergence function, initialized kl_divergence_attributions, and identified {num_latents} latents.")
else:
    num_latents = 0
    print("Crosscoder outputs are empty, cannot determine num_latents.")
````

## Compute KL-Divergence Attribution

### Subtask:
Compute the KL-Divergence attribution for each prompt by comparing activations from `chat_model` (with individual latent interventions) against `base_model` activations at the specified `layer`, ranking latents by negative attribution.


````python
import torch.nn.functional as F
from tqdm.auto import tqdm # Ensure tqdm is imported for progress bars

# Removed: from nnsight.helpers import get_logger
# Removed: nnsight_logger = get_logger("nnsight")
# Removed: nnsight_logger.setLevel("WARNING")

kl_divergence_attributions = []

# Update total_num_latents based on df, as discussed in previous steps
total_num_latents = len(df) # df is the dataframe loaded with latent info
print(f"Total number of latents identified from df: {total_num_latents}")

# Identify the correct latent feature source from crosscoder_model.decoder.weight
# The decoder.weight shape is [2, 73728, 2304]. Assuming index 1 corresponds to chat model features.
latent_feature_source = crosscoder_model.decoder.weight[1] # This should be [73728, 2304]

# Update decoder_size based on the identified latent_feature_source
decoder_size = latent_feature_source.shape[0] # Should be 73728
print(f"Updated decoder size based on identified latent features: {decoder_size}")

print("Computing KL-Divergence attributions...")

# We are only working with the first 5 prompts as per previous steps
for prompt_idx in tqdm(range(len(refusal_prompts[:5])), desc="Processing prompts for attribution"):
    # Re-tokenize for each prompt as nnsight's trace works with fresh tokens
    current_tokens = loaded_chat_model.tokenizer.apply_chat_template(
        [{'role': 'user', 'content': refusal_prompts[prompt_idx]['instruction']}],
        return_tensors="pt"
    ).to(cc_device)

    # Move original activations and log probabilities to the correct device
    original_chat_log_probs = chat_log_probs[prompt_idx].to(cc_device)

    # Dictionary to store attributions for current prompt across all chat-specific latents
    prompt_attributions = {}

    for latent_id in tqdm(chat_specific_latent_ids, leave=False, desc=f"Prompt {prompt_idx} latents"):
        # Check if latent_id is within the bounds of the actual dictionary size
        if latent_id >= decoder_size or latent_id < 0:
            # print(f"Warning: Latent ID {latent_id} out of bounds for dictionary (size {decoder_size}). Skipping.")
            continue

        # Get the feature vector for the specific latent from the identified latent_feature_source
        latent_feature_vector = latent_feature_source[latent_id].to(cc_device)

        # Use nnsight's trace context to run forward pass with intervention
        with loaded_chat_model.trace(current_tokens, disable_tokenizer_output=True) as tracer:
            # Correctly target the hidden states tensor for in-place addition
            # This assumes GemmaDecoderLayer outputs (hidden_states_tuple, past_key_value_states_tuple)
            # where hidden_states_tuple is (hidden_states_tensor,)
            loaded_chat_model.model.layers[layer].output[0][0] += latent_feature_vector.unsqueeze(0).unsqueeze(0).bfloat16()

            # Get the logits from the modified model state
            intervened_logits = loaded_chat_model.lm_head.output.save()

        # Compute log_softmax and move to device for KL-Divergence calculation
        intervened_log_probs = F.log_softmax(intervened_logits.cpu(), dim=-1).to(cc_device)

        # Compute KL-Divergence between original chat log probabilities and intervened log probabilities
        # log_P is original_chat_log_probs (target distribution), log_Q is intervened_log_probs (approximation)
        kl_div = compute_kl_divergence(original_chat_log_probs, intervened_log_probs)

        # Store attribution as negative KL-Divergence, as requested in the task description
        # A larger positive KL-Divergence (more impact/divergence) results in a more negative attribution score.
        prompt_attributions[latent_id] = -kl_div.item()

    kl_divergence_attributions.append({
        'prompt_idx': prompt_idx,
        'prompt_instruction': refusal_prompts[prompt_idx]['instruction'],
        'attributions': prompt_attributions
    })

print("KL-Divergence attribution computation complete.")
````

**Reasoning**:
The KL-Divergence attribution computation is complete. The next step is to aggregate the attribution scores for each chat-specific latent across all processed prompts and rank them to identify the top refusal latents as requested by the task.



````python
from collections import defaultdict

# Aggregate attribution scores for each latent across all prompts
aggregated_attributions = defaultdict(float)
for prompt_result in kl_divergence_attributions:
    for latent_id, attribution_score in prompt_result['attributions'].items():
        aggregated_attributions[latent_id] += attribution_score

# Calculate average attribution for each latent
num_prompts_processed = len(refusal_prompts[:5])
if num_prompts_processed > 0:
    averaged_attributions = {latent_id: score / num_prompts_processed
                             for latent_id, score in aggregated_attributions.items()}
else:
    averaged_attributions = {}

# Rank latents by averaged negative attribution (higher negative value means stronger attribution)
ranked_latents = sorted(averaged_attributions.items(), key=lambda item: item[1], reverse=False) # reverse=False for most negative first

print("Top 10 refusal latents based on KL-Divergence attribution (lower score = more refusal-like):")
for i, (latent_id, score) in enumerate(ranked_latents[:10]):
    print(f"  {i+1}. Latent ID: {latent_id}, Average Negative KL-Divergence Attribution: {score:.6f}")

# Optionally, store the top 10 latent IDs
top_refusal_latent_ids = [latent_id for latent_id, _ in ranked_latents[:10]]
print(f"\nStored top 10 refusal latent IDs: {top_refusal_latent_ids}")
````
